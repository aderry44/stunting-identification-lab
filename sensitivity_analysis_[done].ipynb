#Sc.1. Result Preview

```python
sc1_train_result_df = pd.read_excel(base + 'sc1_train_result.xlsx')
sc1_result_df       = pd.read_excel(base + 'sc1_result.xlsx')

sc1_test_result_df          = sc1_result_df[sc1_result_df.evaluation == "test"]
sc1_test_result_df[metrics] = sc1_test_result_df[metrics] * 100

sc1_summary_test_df = pd.merge(sc1_train_result_df,
                               sc1_test_result_df,
                               on=['model_name', 'scheme', 'model', 'optimizer'],
                               how='inner')

sc1_summary_test_df = sc1_summary_test_df.loc[:,
 ["scheme", "model", "optimizer", "best_epoch", "sensitivity", "specificity", "f2"]]
```

```python
sc1_summary_test_df.round(2)
```

#1\. Wihtin Correlation Sensitivity Analaysis





```python
sensitivity_analysis_df_1 = pd.merge(sc1_summary_test_df[sc1_summary_test_df.scheme == "sc1a"],
                                     sc1_summary_test_df[sc1_summary_test_df.scheme == "sc1b"],
                                     on=['model', 'optimizer'],
                                     how='inner',
                                     suffixes=("_sc1a", "_sc1b")
                                     )

sensitivity_analysis_df_1["f2_diff"]     = sensitivity_analysis_df_1["f2_sc1a"] - sensitivity_analysis_df_1["f2_sc1b"]
sensitivity_analysis_df_1["f2_diff_pct"] = (sensitivity_analysis_df_1["f2_sc1a"] - sensitivity_analysis_df_1["f2_sc1b"]) / sensitivity_analysis_df_1["f2_sc1a"] * 100
```

```python
sensitivity_analysis_df_1.drop(columns=['scheme_sc1a', 'scheme_sc1b']).round(2)
```
